## 2. Values and lists

A **variable** stores a value under a name: `takings = 250`. Use the name later and Python fills in the value.

**Arithmetic:** `+`, `-`, `*` and `/` work as you would expect. Powers use `**`, so `3 ** 2` is `9`.

- **Watch out:** `^` is *not* a power in Python. `3 ^ 2` gives `1`, and there is no error to warn you.
- `/` always gives a decimal: `7 / 2` is `3.5`. `//` drops the remainder: `7 // 2` is `3`.

**f-strings** put values inside text: `print(f"Takings: {takings}")`. The `f` before the quote lets you use `{ }`.

Names are **case-sensitive**: `Total` and `total` are two different variables. `True`, `False` and `None` need capital letters.

A **list** holds several values in order: `week = [250, 310, 180, 400, 520]`.

- Lists **count from 0**. `week[0]` is Monday and `week[4]` is Friday, so `week[1]` is the *second* item.
- `len(week)` counts the items. `sum(week)`, `min(week)` and `max(week)` do what they say.
- The average is `sum(week) / len(week)`.
- `week.append(610)` adds a value to the end. The dot means "do this to the list"; the list itself changes.
- **Watch out:** do not write `week = week.append(610)`. `append` gives back `None`, so `week` would become `None` and your list would be lost.
- `week.index(max(week))` finds the position of the best day, counted from 0.

## 3. Loops, decisions, and pattern 1

In Python, **the indent decides** what is inside a loop or an `if`. A line ending in a colon (`:`) must be followed by at least one indented line, and every indented line belongs to it.

```python
total = 0
for day in week:
    total += day      # indented: runs every time round the loop
print(total)          # not indented: runs once, after the loop
```

Move that `print` in by four spaces and it runs every time instead. You get a different answer and **no error**, which is why indentation matters so much.

- One indent is **4 spaces**. The Tab key types 4 spaces here. Colab may use 2 spaces instead. Both work, as long as every line in one block matches.
- A real tab character looks just like spaces, but Python treats it differently. Mixing the two gives `IndentationError` or `TabError`, usually after pasting code from an email or a PDF. The fix: delete the indent and press Tab again. In this editor, tab characters are highlighted.

**Loops.** `for day in week:` goes through each item. `for i in range(5):` counts 0, 1, 2, 3, 4: `range(5)` stops *before* 5. `range(1, 6)` counts 1 to 5. To add up as you go, start with `total = 0` *before* the loop, then `total += amount` inside it (short for `total = total + amount`).

**Decisions.** `if amount > 300:` … `elif amount > 200:` … `else:`. Python writes `elif`, not `else if`. `==` compares, `=` stores: `if total == 5:` is a question. `and`, `or` and `not` are written as words.

**Step and record.** Each time round a loop, update a value, then `history.append(value)`. Plotting `history` shows how it changed: the same shape as the mortgage section of the Colab notebook.

## 4. Functions and randomness

A **function** is a named recipe. You write it once and use it as often as you like.

```python
def add_interest(balance, rate):
    return balance * (1 + rate)
```

The line starts with `def`, then the function's name, then its **arguments** in brackets, then a colon. The body underneath is indented by 4 spaces. Defining a function does nothing on its own: it only runs when you **call** it, for example `add_interest(1000, 0.005)`.

An argument can have a **default value**: `def add_interest(balance, rate=0.005):`. Now `add_interest(1000)` works too, and uses 0.005.

**`print` versus `return`.** `return` hands a value back, so you can store it: `new = add_interest(1000, 0.005)`. `print` only shows a value on the screen. A function without `return` gives back `None`, so after `x = f(5)`, `x` is `None`. This is the most common mistake in this block.

A function can read a variable defined outside it. If that variable changes later, the function's answer changes silently. It is safer to pass the value in as an argument.

**Randomness.** Start with `import random`.

- `random.random()` gives a decimal between 0 and 1 (it can be 0, but never 1).
- `random.randint(1, 6)` rolls a dice. **Both ends are included**, unlike `range(1, 6)`, which stops at 5.
- `random.choice(week)` picks one item from a list.
- `random.seed(1)` gives the same "random" numbers every run, so results can be checked. The tutorial does this for you when it checks your code.

**Wrap and repeat:** put one run of a simulation in a function, call it many times in a loop, and count how often something happens. That share is a probability, estimated by simulation.

## 5. Pattern 3: sweep and compare

So far you have run a function once. Real questions need more than one run. **Which setting works best?** To answer that, you *sweep*: run the same function for several settings and compare the results.

The pattern has three steps:

1. **Make a list of settings**, for example `orders = [20, 30, 40, 50, 60]`.
2. **Run the function for each setting** in a `for` loop, and store each answer with `results.append(...)`. Start with an empty list, `results = []`, *before* the loop.
3. **Pick the best.** `max(results)` gives the best *value*. To find the *setting* that gave it, use its position: `orders[results.index(max(results))]`.

A plot makes the comparison easy to see: `plt.plot(orders, results)` puts the settings along the bottom and the results up the side.

Two things to notice when you look at the curve:

- **The top is often flat.** Several settings give almost the same result, so being roughly right is usually enough. Don't agonise over the exact peak.
- **One run is not a finding.** When a function uses random numbers, each run gives a slightly different answer. Average over many days, or many runs, before you trust a difference. The spread across runs matters as much as any single number.

This is the same shape as the fish-and-chip shop section in the organiser's notebook: a model, a list of settings, a loop, and a plot.

### Sweep and compare: the bakery

A bakery orders loaves each morning. Each loaf costs £1 and sells for £3. Demand is random; unsold loaves are wasted.

`profit(order)` simulates 200 days. Each day it calls `random.randint(20, 60)` **once** for that day's demand, and it returns the average daily profit.

Look at the top of the curve: is it sharp or flat?

````python
import random

price = 3
cost = 1

def profit(order):
    total = 0
    for day in range(200):
        demand = random.randint(20, 60)
        sold = min(order, demand)
        total += sold * price - order * cost
    return total / 200

orders = [20, 30, 40, 50, 60]
# 1. make an empty list called profits
# 2. loop over orders and append profit(order)
# 3. best_order = ...

import matplotlib.pyplot as plt
# 4. plt.plot(orders, profits) and so on
````
